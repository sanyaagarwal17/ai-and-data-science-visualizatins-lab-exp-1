```python
from pandas import Series

# Create a Series
sales = Series( [300, 750, 450, 900], index=['Pen', 'Bag', 'Book', 'Laptop'])
# Display the complete Series
print("Sales Series:")
print(sales)
# Display values
print("\nData Values:")
print(sales.values)
# Display index
print("\nIndex:")
print(sales.index)
# Select one label
print("\nSales of Bag:")
print(sales['Bag'])
# Filter sales above 500
print("\nSales above 500:")
print(sales[sales > 500])
# Check whether a label exists
print("\nIs Book present?")
print('Book' in sales)
# Convert Series to dictionary
print("\nDictionary:")
print(sales.to_dict())
```

```python
import pandas as pd
from pandas import Series
data = {'A': 15, 'B': 35, 'C': 25}
s = Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'
print("Complete Series:")
print(s)
print("\nValues greater than 20:")
print(s[s > 20])
print("\nNon-missing values:")
print(pd.notnull(s))
```
