```python
import numpy as np
np.random.seed(42)
marks = np.random.randint(20, 101, size=(12, 4))
print("Marks:")
print(marks)
student_avg = np.mean(marks, axis=1)
test_mean = np.mean(marks, axis=0)
test_std = np.std(marks, axis=0)
print("Student averages:")
print(np.round(student_avg, 2))
print("Test means:", np.round(test_mean, 2))
print("Test standard deviations:", np.round(test_std, 2))
passed_all = np.all(marks >= 40, axis=1)
print("Students passing all tests:", np.sum(passed_all))
print("Student numbers passing all tests:", np.where(passed_all)[0] + 1)
high_avg = np.where(student_avg >= 75)[0] + 1
print("Students averaging at least 75:", high_avg)
most_variable_test = np.argmax(test_std) + 1
print("Test with greatest variation:", most_variable_test)
```

```python
import numpy as np
a = np.arange(21, 46).reshape(5, 5)
print("Original matrix:")
print(a)
central = a[1:4, 1:4]
print("Central 3x3 matrix:")
print(central)
transposed = central.T
print("Transpose:")
print(transposed)
product = central @ transposed
print("Matrix product:")
print(product)
row_sums = np.sum(product, axis=1)
print("Row-wise sums:", row_sums)
print("Maximum row sum:", np.max(row_sums))
print("Positions tied at maximum:", np.where(row_sums == np.max(row_sums))[0])
```

```python
import numpy as np
quantity = np.array([
    [20, 15, 30, 10],
    [25, 18, 25, 12],
    [15, 20, 35, 15],
    [30, 25, 20, 18],
    [22, 30, 28, 20],
    [28, 22, 32, 16]])
original_prices = np.array([150, 240, 90, 320])
daily_revenue = np.sum(quantity * original_prices, axis=1)
product_quantity = np.sum(quantity, axis=0)
product_revenue = np.sum(quantity * original_prices, axis=0)
print("Daily revenue:", daily_revenue)
print("Total quantity per product:", product_quantity)
print("Revenue per product:", product_revenue)
print("Total revenue:", np.sum(daily_revenue))
print("Highest-revenue day:", np.argmax(daily_revenue) + 1)
print("Best-selling product:", np.argmax(product_quantity) + 1)
new_prices = np.where(
    product_quantity > 100,
    original_prices * 0.90,
    original_prices)
print("Original prices:", original_prices)
print("Reduced prices:", new_prices)
new_daily_revenue = np.sum(quantity * new_prices, axis=1)
print("New daily revenue:", new_daily_revenue)
print("New total revenue:", np.sum(new_daily_revenue))
```

```python
import numpy as np
np.random.seed(42)
data = np.random.normal(50, 12, 100)
cleaned = np.where(data < 30, 30 ,np.where(data > 70, 70, data))
print("Values outside 30-70:", np.sum((data < 30) | (data > 70)))
print("\nOriginal statistics:")
print("Mean:", np.mean(data))
print("Variance:", np.var(data))
print("Standard deviation:", np.std(data))
print("Minimum:", np.min(data))
print("Maximum:", np.max(data))
print("\nCleaned statistics:")
print("Mean:", np.mean(cleaned))
print("Variance:", np.var(cleaned))
print("Standard deviation:", np.std(cleaned))
print("Minimum:", np.min(cleaned))
print("Maximum:", np.max(cleaned))
```
