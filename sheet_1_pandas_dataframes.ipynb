```python
import numpy as np
import pandas as pd
df = pd.DataFrame({'Name': ['Aman', 'Riya', 'Rahul', 'Priya', 'Neha']})
df['SerialNo'] = np.arange(5)
print(df)
```

```python
movie_df = pd.DataFrame({'Movie': ['Dangal', 'Pathaan', 'Jawan'],'Year': [2016, 2023, 2023],'Rating': [8.5, 7.0, 7.5]})

print(movie_df)
```

```python
new_movie_df = movie_df[['Movie', 'Rating']].copy()
new_movie_df['Director'] = ['Nitesh Tiwari','Siddharth Anand','Atlee']
print(new_movie_df)
```

```python
cities = pd.Series(['Delhi', 'Jaipur'], index=[0, 3])
print(cities)
```

```python
branch_df = pd.DataFrame(index=range(4))
branch_df['Branch'] = cities
print(branch_df)
```

```python
employee_df = pd.DataFrame({'Employee': ['Aman', 'Riya', 'Rahul', 'Priya', 'Neha'],'Department': ['IT', 'HR', 'Sales', 'IT', 'Finance'],'Experience': [2, 3, 1, 4, 2]})
print(employee_df[['Employee', 'Experience']])
```

```python
product_df = pd.DataFrame({'Product': ['Laptop', 'Mobile', 'Tablet'],'Category': ['Electronics', 'Electronics', 'Electronics'],'Price': [50000, 20000, 15000]})
new_product_df = product_df[['Price', 'Product', 'Category']]
print(new_product_df)
```

```python
data = {'Name': ['Aman', 'Riya', 'Rahul'],'Age': [20, 21, 19],'City': ['Delhi', 'Meerut', 'Jaipur']}
df = pd.DataFrame(data)
new_df = pd.DataFrame( df,columns=['Name', 'Age', 'City', 'Marks'])
print(new_df)
```

```python
marks_df = pd.DataFrame({'Name': ['Aman', 'Riya', 'Rahul'],'Total Marks': [80, 90, 75]})
print(marks_df['Total Marks'])
```

```python
course_df = pd.DataFrame({'Student': ['Aman', 'Riya', 'Rahul', 'Priya', 'Neha'],'Course': ['Python', 'Java', 'C++', 'Python', 'Java'],'Score': [85, 90, 78, 88, 92]})
print(course_df.columns)
print(course_df['Student'])
new_df = course_df[['Student', 'Score']]
new_df['Batch'] = 'BCA'
print(new_df)
```

```python

travel_df = pd.DataFrame({'Place': ['Delhi', 'Agra', 'Jaipur', 'Mumbai', 'Pune']})
travel_df['Code'] = np.arange(5)
print(travel_df)
codes = pd.Series([100, 400], index=[1, 4])
travel_df['Code'] = codes
print(travel_df)
```

```python
school_df = pd.DataFrame({ 'Student': ['Aman', 'Riya', 'Rahul', 'Priya'],'Class': [10, 10, 12, 12],'Section': ['A', 'B', 'A', 'B']})
new_df = pd.DataFrame( school_df, columns=['Student', 'Class', 'Section', 'House'])
new_df['House'] = 'Blue House'
print(new_df)
```

```python
inventory_df = pd.DataFrame({'Item': ['Pen', 'Book', 'Pencil', 'Bag', 'Eraser']})
inventory_df['Stock'] = np.arange(5)
print(inventory_df)
stock = pd.Series([50, 100], index=[1, 3])
inventory_df['Stock'] = stock
print(inventory_df)
inventory_df = inventory_df.drop('Stock', axis=1)
print(inventory_df)
```
